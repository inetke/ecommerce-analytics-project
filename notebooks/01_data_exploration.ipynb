# E-commerce Data Exploration

This notebook explores the Olist Brazilian E-Commerce dataset to understand its structure, relationships and key variables before performing the SQL and business analysis.

The goal of this initial exploration is to understand the available data and identify the tables and fields needed to analyze sales, customers, product performance and delivery experience.

```python
import pandas as pd
```

## Orders

```python
orders = pd.read_csv("../data/raw/olist_orders_dataset.csv")
```

```python
orders.head()
```

```python
orders.columns
```

### Orders table

Each row represents an e-commerce order.

- `order_id`: Unique identifier for each order.
- `customer_id`: Identifier of the customer associated with the order.
- `order_status`: Current/final status of the order.
- `order_purchase_timestamp`: Date and time when the purchase was placed.
- `order_approved_at`: Date and time when the order/payment was approved.
- `order_delivered_carrier_date`: Date and time when the order was handed over to the carrier.
- `order_delivered_customer_date`: Actual delivery date to the customer.
- `order_estimated_delivery_date`: Estimated delivery date.

These timestamps can later be used to analyze delivery times, delays and their potential relationship with customer satisfaction.

## Customers

```python
customers = pd.read_csv("../data/raw/olist_customers_dataset.csv")

customers.head()
```

```python
customers.columns
```

```python
customers["customer_id"].nunique()
```

```python
customers["customer_unique_id"].nunique()
```

### Customers table

Each row represents a customer record associated with an order.

- `customer_id`: Identifier associated with the customer for a specific order. The same person may have different `customer_id` values across different orders.
- `customer_unique_id`: Unique identifier for the actual customer. This allows us to identify multiple orders made by the same person.
- `customer_zip_code_prefix`: Prefix of the customer's ZIP code.
- `customer_city`: Customer's city.
- `customer_state`: Customer's state in Brazil.

The distinction between `customer_id` and `customer_unique_id` is important for customer analysis. `customer_id` is used to connect the customers table with the orders table, while `customer_unique_id` should be used to identify unique and repeat customers.

This table will later be used to analyze customer distribution, repeat purchases and customer retention.

## Order Items

```python
order_items = pd.read_csv("../data/raw/olist_order_items_dataset.csv")
```

```python
order_items.head(15)
```

```python
order_items.columns
```

### Order Items table

Each row represents an individual item within an order. A single order can contain multiple items.

- `order_id`: Identifier of the order. Used to connect order items with the orders table.
- `order_item_id`: Sequential number identifying each item within an order.
- `product_id`: Identifier of the product purchased.
- `seller_id`: Identifier of the seller who sold the product.
- `shipping_limit_date`: Deadline for the seller to hand the item over for shipping.
- `price`: Price of the individual item.
- `freight_value`: Shipping cost associated with the item.

An `order_id` can appear multiple times when an order contains more than one item. Therefore, the number of rows in this table does not represent the number of orders.

This table will be used to analyze product sales, revenue, freight costs and seller performance.

## Payments

```python
payments = pd.read_csv("../data/raw/olist_order_payments_dataset.csv")
```

```python
payments.head()
```

```python
payments.columns
```

### Payments table

Each row represents a payment transaction associated with an order. An order may contain more than one payment transaction.

- `order_id`: Identifier of the order. Used to connect payments with the Orders table.
- `payment_sequential`: Sequential number of the payment transaction within an order.
- `payment_type`: Payment method used by the customer.
- `payment_installments`: Number of installments selected for the payment.
- `payment_value`: Total value of the payment transaction.

The same `order_id` may appear multiple times when an order uses more than one payment transaction.

It is important to distinguish `payment_value` from the `price` field in the Order Items table. Their relationship will be investigated before defining the project's revenue metrics.

This table will be used to analyze payment methods, installments and customer payment values.

## Reviews

```python
reviews = pd.read_csv("../data/raw/olist_order_reviews_dataset.csv")
```

```python
reviews.head()
```

```python
reviews.columns
```

### Reviews table

Each row represents a customer review associated with an order.

- `review_id`: Unique identifier for the review.
- `order_id`: Identifier of the order associated with the review.
- `review_score`: Customer rating, ranging from 1 to 5.
- `review_comment_title`: Optional title provided with the review.
- `review_comment_message`: Optional written feedback from the customer.
- `review_creation_date`: Date when the review survey was created.
- `review_answer_timestamp`: Date and time when the customer submitted the review.

Customers can provide a review score without leaving a written comment, so `review_score` will be the primary variable used to measure customer satisfaction.

The `order_id` connects this table with the Orders table.

This table will be used to analyze customer satisfaction and investigate whether delivery performance is associated with lower or higher review scores.

### Geolocation table

Each row represents a geographical location associated with a Brazilian ZIP code prefix.

- `geolocation_city`: City associated with the location.
- `geolocation_state`: Brazilian state associated with the location.

A ZIP code prefix can appear multiple times with slightly different coordinates. Therefore, this table should not be joined directly with customer or seller data without first handling duplicate ZIP code prefixes.

This table can later be used to support geographical analysis and map visualizations.

### Orders — Data Quality Findings

The Orders table contains 99,441 records and 8 columns.

- `order_id` is unique, with no duplicated order identifiers.
- No fully duplicated rows were found.
- Several order lifecycle date fields contain missing values.
- Most missing delivery dates are consistent with orders that were canceled, unavailable, shipped, processing or in earlier stages of the order lifecycle.
- However, 8 orders are marked as `delivered` while `order_delivered_customer_date` is missing. These records represent a potential data quality inconsistency.
- The date columns were initially loaded as strings and will need to be converted to datetime before calculating delivery times and delays.

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

orders[date_columns] = orders[date_columns].apply(pd.to_datetime)
```

```python
orders["order_status"].value_counts(normalize=True) * 100
```

#### Order Status Distribution

Approximately 97% of orders are marked as `delivered`, while the remaining orders are distributed across statuses such as shipped, canceled, unavailable, invoiced and processing.

Because order status may affect whether an order should be included in sales and revenue calculations, the treatment of non-delivered orders will be evaluated before defining the final business KPIs.
